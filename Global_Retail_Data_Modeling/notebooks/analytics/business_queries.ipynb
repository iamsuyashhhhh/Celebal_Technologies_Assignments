```sql
%sql
SELECT customer_sk, SUM(sales_amount) Revenue 
FROM gold.fact_sales
GROUP BY customer_sk
ORDER BY Revenue DESC
LIMIT 10;
```

```sql
%sql
SELECT product_sk, SUM(sales_amount) Revenue FROM gold.fact_sales
GROUP BY product_sk
ORDER BY Revenue DESC
LIMIT 10;
```

```sql
%sql
SELECT p.product_line, SUM(f.sales_amount) Revenue FROM gold.fact_sales f
JOIN gold.dim_product p ON f.product_sk=p.product_sk
GROUP BY p.product_line
ORDER BY Revenue DESC;
```
