```sql
%sql
SELECT SUM(sales_amount) AS total_revenue FROM gold.fact_sales;
```

```sql
%sql
SELECT COUNT(*) AS total_orders FROM gold.fact_sales;
```

```sql
%sql
SELECT COUNT(*) AS total_customers FROM gold.dim_customer;
```

```sql
%sql
SELECT
    c.first_name, c.last_name, SUM(f.sales_amount) AS revenue
FROM gold.fact_sales f
JOIN gold.dim_customer c ON f.customer_sk = c.customer_sk
GROUP BY c.first_name, c.last_name
ORDER BY revenue DESC
LIMIT 10;
```

```sql
%sql
SELECT
    p.product_name, SUM(f.quantity) AS total_quantity
FROM gold.fact_sales f
JOIN gold.dim_product p ON f.product_sk = p.product_sk
GROUP BY p.product_name
ORDER BY total_quantity DESC;
```
